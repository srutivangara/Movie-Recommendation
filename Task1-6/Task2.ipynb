```python
import pandas as pd
import re
import string
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

df = pd.read_csv("tmdb_ds.csv")

print(df.head())
print(df.columns.tolist())
```

```python
df.columns.tolist()
```

```python
df["title"] = df["title"].fillna("")

print("Missing descriptions:", df["title"].isnull().sum())
```

```python
stop_words = set(ENGLISH_STOP_WORDS)

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    words = text.split()
    words = [word for word in words if word not in stop_words]

    return " ".join(words)
```

```python
df["clean_text"] = df["title"].apply(clean_text)
df[["title", "clean_text"]].head()
print("Dataset shape:", df.shape)
print("\nMissing values in clean_text:", df["clean_text"].isnull().sum())

print("\nOriginal description:")
print(df["title"].iloc[0])

print("\nCleaned description:")
print(df["clean_text"].iloc[0])
```
